# Player Detection Testing

This notebook is for testing and experimenting with player detection on your football game film using YOLOv8.

## Setup

```python
import sys
sys.path.append('..')

from src.detection.player_detector import PlayerDetector
import cv2
from pathlib import Path
```

## Initialize Detector

This will download the YOLO model on first run (about 6MB for yolov8n).

```python
# Initialize the detector (this will download the YOLO model on first run)
detector = PlayerDetector(config_path='../config.yaml')
print("Detector initialized successfully!")
```

## Test on a Single Frame

Let's first test detection on a single frame to see if it's working.

```python
# Load a test frame from your video
video_path = "../data/raw/sample_play.mp4"

if Path(video_path).exists():
    cap = cv2.VideoCapture(video_path)
    ret, frame = cap.read()
    cap.release()
    
    if ret:
        # Run detection on this single frame
        results = detector.model(
            frame,
            conf=detector.config['model']['confidence_threshold'],
            device=detector.device,
            verbose=True
        )
        
        # Draw detections
        frame_with_boxes = detector._draw_detections(frame, results[0])
        
        # Display the number of detections
        print(f"Detected {len(results[0].boxes)} objects")
        
        # Show the frame (this will open a window)
        cv2.imshow('Test Frame', frame_with_boxes)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Could not read frame from video")
else:
    print(f"Video file not found: {video_path}")
```

## Process Full Video

Once the single frame test works, process the entire video.

```python
# Process the full video
output_path = detector.detect_players(video_path)
print(f"Done! Output saved to: {output_path}")
```
